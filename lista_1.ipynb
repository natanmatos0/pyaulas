```python
#terceira questao
l = [28, 1, 19, 8, 12, 11, 36, 17, 30, 45]
n = len(l)
for i in range(0, n, 1): #a variavel i percorre todas as posiçõe1
  for j in range(i - 1, n, 1): #a variavel j percorre todas as posições a partir da seguinte do i
    if l[i] < l[j]: #quando o elemento na posição i for maior que na posição j
      aux = l[i] #a variavel aux recebe o valor na posição i
      l[i] = l[j] #o elemento na posição i troca com o da posição j
      l[j] = aux #o elemnto que estava na posiçao j vai pra posição i
print(l[len(l) - 1])
```

```python
#decima primeira questao
l = [29, 29, 16, 10, 22, 49, 9, 21, 6, 18]
nl = []
for numero in l:
    repetido = False
    i = 0
    while i < len(nl):
        if nl[i] == numero:
            repetido = True
            break
        i += 1

    if not repetido:
        nl = nl + [numero]

print(nl)
```

```python
#decima segunda questao
l = [29, 15, 16, 10, 22, 49, 9, 21, 6, 18]
maior = l[0]
segundomaior = l[0]
for num in l:
    if num > maior and num < segundomaior:
        maior = num

print(maior)
```

```python
#decima terceira questao
def bubbleSort(l):
  aux = []
  n = len(l)
  for i in range(0, n , 1):
    for j in range(i + 1, n, 1):
      if l[i] > l[j]:
        aux = l[i]
        l[i] = l[j]
        l[j] = aux
l = [29, 15, 16, 10, 22, 49, 9, 21, 6, 18]
bubbleSort(l)
print(f"A lista em ordem crescente é: {l}")
```
